```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
def plot_model_performance_updated(cohort_df, cohort_name, target_name, metrics = ["ROC", "PR", "F1"] ):
    # Define feature order and colors/markers for metrics
    feature_order = ["X", "M", "D", "MD", "XM", "XD", "XMD"]
    metric_colors = {
        "ROC": "#1f77b4",  # blue
        "PR": "#ff7f0e",  # orange
        "F1": "#2ca02c",  # green
        "Recall": "#d62728",  # red
    }
    metric_markers = {
        "ROC": "D",
        "PR": "o",
        "F1": "s",
        "Recall": "^",
    }  # diamond = "D", circle = "o"
    #metrics = ["ROC", "PR", "F1"], "Recall"]

    models_dict = {
        "LR": "Logistic Regression",
        "RF": "Random Forest",
        "GB": "Gradient Boosting",
        "XGB": "XGBoost",
        "CB": "CatBoost",
    }

    #models = cohort_df["Model"].unique()
    models = ["LR", "RF", "GB", "XGB", "CB"]
    fig, axes = plt.subplots(1, len(models), figsize=(3.5 * len(models), 8), sharey=True)  ### CHANGED
    
    # Iterate over models
    for i, model in enumerate(models):   ### CHANGED
        ax = axes[i]                     ### CHANGED
        
        model_data = cohort_df[cohort_df["Model"] == model].copy()
        model_data["Features"] = pd.Categorical(
            model_data["Features"], categories=feature_order, ordered=True
        )
        model_data = model_data.sort_values("Features")

        # Plot each metric - SWAP X AND Y AXES
        for metric in metrics:
            ax.errorbar(
                model_data["Features"],  # X-axis now features
                model_data[f"{metric}_mean"],  # Y-axis now metric values
                yerr=model_data[f"{metric}_std"],  # yerr instead of xerr
                fmt=metric_markers[metric],
                markersize=10,
                color=metric_colors[metric],
                ecolor=metric_colors[metric],
                elinewidth=2,
                capsize=6,
                label=metric,
                linestyle="--",
                linewidth=1,
                markeredgecolor="white",
                markeredgewidth=1.5,
            )

        # Performance thresholds - NOW HORIZONTAL LINES
        ax.axhspan(
            0.8, 0.9, color="#2E8B57", alpha=0.15, label="Good (0.8–0.9)"
        )  # Sea Green
        ax.axhspan(
            0.7, 0.8, color="#FFA500", alpha=0.15, label="Fair (0.7–0.8)"
        )  # Orange
        ax.axhline(0.8, color="#2E8B57", linestyle="--", linewidth=1.5, alpha=0.7)
        ax.axhline(0.7, color="#FFA500", linestyle="--", linewidth=1.5, alpha=0.7)

        # Set major and minor ticks on Y-axis instead of X-axis
        major_ticks = np.arange(0.0, 0.91, 0.1)
        ax.set_yticks(major_ticks)

        minor_ticks = np.arange(0.01, 0.91, 0.01)
        ax.set_yticks(minor_ticks, minor=True)

        # Configure Y-axis ticks instead of X-axis
        ax.tick_params(axis="y", which="major", length=10, labelsize=14)
        ax.tick_params(axis="y", which="minor", length=5, labelsize=14)

        # Rotate x-axis labels for better readability
        ax.tick_params(axis="x", rotation=0)

        # Titles and labels - SWAPPED
        ax.set_title(f"{models_dict[model]}", fontsize=16)
        ax.set_ylabel("Score (Mean ± Standard Deviation)", fontsize=16)
        ax.set_xlabel("Feature Set", fontsize=16)
        low, high = (
            cohort_df[[f"{m}_mean" for m in metrics]].min().min(),
            cohort_df[[f"{m}_mean" for m in metrics]].max().max(),
        )
        ax.set_ylim(max(0.0, low - 0.05), min(0.9, high + 0.05))

        # ax.set_ylim(0.0, 0.9)  # ylim instead of xlim

        # Grid on Y-axis instead of default (both)
        ax.grid(
            True, axis="y", color="lightgray", linestyle="--", linewidth=0.7, alpha=0.7
        )
        if i == len(models)-1:
            ax.legend(fontsize=14)

    fig.tight_layout()

    # Save as PDF
    fig.savefig(f"{cohort_name}_{target_name}_{model}_metrics_updated.pdf", format="pdf", bbox_inches="tight",)
    
    # also save as PNG in high resolution
    dir = Path.cwd().parent.parent / "results"
    fig.savefig(dir / f"{cohort_name}_{target_name}_{model}_metrics.png", format="png", dpi=300)
    #plt.close(fig)
    plt.show()

plot_model_performance_updated(apl_df, "APL", "Clinical_Target")
```
